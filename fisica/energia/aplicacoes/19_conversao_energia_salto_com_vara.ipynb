# Conversão de Energia Cinética em Potencial Gravitacional
### Trabalho, energia e suas transformações

**Objetivo:** Calcular a altura máxima quando uma fração da energia cinética é convertida em energia potencial gravitacional.

### Como usar este notebook

Execute as células na ordem, de cima para baixo. As células Markdown apresentam os conceitos; as células de código fazem as contas e geram gráficos com **NumPy** e **Matplotlib**.

> **Convenção:** neste material, usamos principalmente $W$ para trabalho, medido em joules (J). Algumas fontes usam $\tau$, mas esse símbolo também é muito utilizado para torque. O trabalho é uma grandeza escalar.

## 1. Conceito principal

Em uma conversão idealizada, uma fração $\eta$ da energia cinética inicial vira energia potencial gravitacional:

$$
mgh_{\max}=\eta\frac{mv^2}{2}
$$

Cancelando a massa:

$$
\boxed{h_{\max}=\eta\frac{v^2}{2g}}
$$

A eficiência $\eta$ deve ser escrita como fração: $80\%=0{,}80$.

## 2. Exemplo resolvido passo a passo

Um atleta alcança $v=11\,\mathrm{m/s}$ e converte $80\%$ de sua energia cinética em energia potencial. Considere $g=10\,\mathrm{m/s^2}$:

$$
h_{\max}=0{,}80\cdot\frac{11^2}{2\cdot10}=0{,}80\cdot6{,}05=\boxed{4{,}84\,\mathrm{m}}
$$

Esse é um modelo de conversão de energia; um salto com vara real tem outras etapas e perdas, além da dinâmica da vara e do atleta.

## 3. Conferindo o cálculo com Python

A próxima célula implementa a fórmula usando as grandezas e unidades indicadas no exemplo.

```python
v = 11.0
eta = 0.80
g = 10.0
h_max = eta * v**2 / (2*g)
fracao_dissipada = 1 - eta
print(f"Eficiência de conversão: {eta:.0%}")
print(f"Altura estimada: {h_max:.2f} m")
print(f"Fração não convertida em energia potencial: {fracao_dissipada:.0%}")
```

## 4. Visualização

O gráfico abaixo ajuda a interpretar a relação física, além de conferir os valores numéricos.

```python
import matplotlib.pyplot as plt

categorias = ["Convertida em Ep", "Não convertida em Ep"]
fracoes = [eta*100, (1-eta)*100]
plt.figure(figsize=(7.5, 4.5))
plt.bar(categorias, fracoes)
plt.ylabel("Fração da energia cinética inicial (%)")
plt.title("Distribuição da energia no modelo de eficiência")
plt.ylim(0, 100)
plt.grid(axis="y", alpha=0.3)
plt.show()
```

## 5. Como interpretar o resultado

A massa não aparece na fórmula final porque ela se cancela. Para a mesma velocidade e gravidade, a altura prevista cresce proporcionalmente à eficiência. O modelo pressupõe que a fração convertida vira energia potencial gravitacional.

## 6. Exercício de fixação

Um corpo a $8\,\mathrm{m/s}$ converte $75\%$ da energia cinética em energia potencial gravitacional. Use $g=10\,\mathrm{m/s^2}$ para estimar a altura.

Tente prever o resultado antes de executar a célula seguinte.

```python
v_ex, eta_ex, g_ex = 8.0, 0.75, 10.0
h_ex = eta_ex * v_ex**2 / (2*g_ex)
print(f"Altura estimada = {h_ex:.2f} m")
```

## 7. Resumo

- $mgh=\eta\frac12mv^2$.
- $h=\eta v^2/(2g)$; a massa se cancela.
- Interprete a eficiência como a fração da energia inicial que vai para a forma final estudada.
