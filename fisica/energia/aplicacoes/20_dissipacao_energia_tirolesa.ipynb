# Dissipação de Energia Mecânica em uma Tirolesa
### Trabalho, energia e suas transformações

**Objetivo:** Calcular a velocidade final considerando que uma parte da energia mecânica inicial é dissipada por atrito.

### Como usar este notebook

Execute as células na ordem, de cima para baixo. As células Markdown apresentam os conceitos; as células de código fazem as contas e geram gráficos com **NumPy** e **Matplotlib**.

> **Convenção:** neste material, usamos principalmente $W$ para trabalho, medido em joules (J). Algumas fontes usam $\tau$, mas esse símbolo também é muito utilizado para torque. O trabalho é uma grandeza escalar.

## 1. Conceito principal

Quando existe atrito, a energia mecânica não precisa permanecer constante: parte dela é transformada em energia interna, calor, som ou deformações. Se uma fração $p$ da energia inicial é perdida, a fração restante é $1-p$.

Para uma pessoa que parte do repouso a uma altura $h$ e chega ao nível de referência, se uma fração $\eta$ da energia potencial inicial virar energia cinética:

$$
\frac12mv^2=\eta mgh\quad\Rightarrow\quad\boxed{v=\sqrt{2\eta gh}}
$$

A massa se cancela nesse modelo.

## 2. Exemplo resolvido passo a passo

Uma pessoa de $50\,\mathrm{kg}$ parte do repouso de uma altura de $5\,\mathrm{m}$. Perde-se $36\%$ da energia mecânica por atrito, portanto restam $64\%=0{,}64$. Use $g=10\,\mathrm{m/s^2}$.

$$
\frac12mv^2=0{,}64mgh
$$
$$
v^2=2\cdot0{,}64\cdot10\cdot5=64\Rightarrow\boxed{v=8\,\mathrm{m/s}}
$$

Assume-se que a pessoa termina no nível de referência e que os $64\%$ restantes correspondem à energia cinética final.

## 3. Conferindo o cálculo com Python

A próxima célula implementa a fórmula usando as grandezas e unidades indicadas no exemplo.

```python
m = 50.0       # kg
g = 10.0       # m/s²
h = 5.0        # m
perda = 0.36
fracao_restante = 1 - perda
E_inicial = m*g*h
E_dissipada = perda * E_inicial
Ec_final = fracao_restante * E_inicial
v_final = (2*Ec_final/m)**0.5

print(f"Energia mecânica inicial: {E_inicial:.0f} J")
print(f"Energia dissipada:        {E_dissipada:.0f} J")
print(f"Energia cinética final:   {Ec_final:.0f} J")
print(f"Velocidade final:         {v_final:.1f} m/s")
```

## 4. Visualização

O gráfico abaixo ajuda a interpretar a relação física, além de conferir os valores numéricos.

```python
import matplotlib.pyplot as plt

nomes = ["Dissipada", "Convertida em Ec"]
energias = [E_dissipada, Ec_final]
plt.figure(figsize=(7.5, 4.5))
plt.bar(nomes, energias)
plt.ylabel("Energia (J)")
plt.title("Destino da energia potencial inicial")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

## 5. Como interpretar o resultado

Dos $2500\,\mathrm{J}$ de energia potencial inicial, $900\,\mathrm{J}$ são dissipados e $1600\,\mathrm{J}$ permanecem como energia cinética final. Ao considerar a dissipação, a velocidade final é menor do que no caso ideal sem atrito, que daria $10\,\mathrm{m/s}$.

## 6. Exercício de fixação

Uma pessoa parte do repouso de $8\,\mathrm{m}$ de altura. Se $25\%$ da energia inicial é dissipada, qual é a velocidade final? Use $g=10\,\mathrm{m/s^2}$.

Tente prever o resultado antes de executar a célula seguinte.

```python
h_ex, perda_ex, g_ex = 8.0, 0.25, 10.0
eta_ex = 1 - perda_ex
v_ex = (2*eta_ex*g_ex*h_ex)**0.5
print(f"Velocidade final = {v_ex:.2f} m/s")
```

## 7. Resumo

- Com dissipação, a energia mecânica diminui.
- Fração restante: $\eta=1-p$.
- Para partida do repouso e chegada ao nível de referência: $v=\sqrt{2\eta gh}$.
