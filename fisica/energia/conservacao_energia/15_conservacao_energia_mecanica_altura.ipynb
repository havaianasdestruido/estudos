# Conservação da Energia Mecânica e Perda por Atrito
### Trabalho, energia e suas transformações

**Objetivo:** Encontrar uma altura usando a conservação de energia mecânica em um sistema sem atrito.

### Como usar este notebook

Execute as células na ordem, de cima para baixo. As células Markdown apresentam os conceitos; as células de código fazem as contas e geram gráficos com **NumPy** e **Matplotlib**.

> **Convenção:** neste material, usamos principalmente $W$ para trabalho, medido em joules (J). Algumas fontes usam $\tau$, mas esse símbolo também é muito utilizado para torque. O trabalho é uma grandeza escalar.

## 1. Conceito principal

Quando somente forças conservativas realizam trabalho, a energia mecânica se conserva:

$$
E_{mA}=E_{mB}
$$

Para um corpo em campo gravitacional:

$$
 mgh_A+\frac{mv_A^2}{2}=mgh_B+\frac{mv_B^2}{2}
$$

Podemos dividir toda a equação pela massa, pois ela aparece em todos os termos. **Atenção:** embora o título da linha original mencione perda por atrito, o exemplo numérico fornecido não inclui atrito; ele é um caso de conservação ideal. Quando há atrito, uma forma de escrever o balanço é $E_{mB}=E_{mA}+W_{\text{não conservativo}}$. Se o trabalho do atrito é negativo, a energia mecânica diminui. A dissipação também é calculada numericamente no notebook de tirolesa.

## 2. Exemplo resolvido passo a passo

Dados: $g=10\,\mathrm{m/s^2}$, $v_A=0$, $h_B=2\,\mathrm{m}$, $v_B=10\,\mathrm{m/s}$. Determine $h_A$.

$$
gh_A+\frac{v_A^2}{2}=gh_B+\frac{v_B^2}{2}
$$
$$
h_A= h_B+\frac{v_B^2-v_A^2}{2g}=2+\frac{100}{20}=\boxed{7\,\mathrm{m}}
$$

## 3. Conferindo o cálculo com Python

A próxima célula implementa a fórmula usando as grandezas e unidades indicadas no exemplo.

```python
g = 10.0
v_A = 0.0
h_B = 2.0
v_B = 10.0
h_A = h_B + (v_B**2 - v_A**2) / (2*g)
print(f"Altura do ponto A: h_A = {h_A:.1f} m")
# Verificação da conservação por unidade de massa (J/kg)
E_por_massa_A = g*h_A + 0.5*v_A**2
E_por_massa_B = g*h_B + 0.5*v_B**2
print(f"E_m/m em A = {E_por_massa_A:.1f} J/kg")
print(f"E_m/m em B = {E_por_massa_B:.1f} J/kg")
```

## 4. Visualização

O gráfico abaixo ajuda a interpretar a relação física, além de conferir os valores numéricos.

```python
import matplotlib.pyplot as plt

labels = ["A: início", "B: final"]
Ep_por_massa = [g*h_A, g*h_B]
Ec_por_massa = [0.5*v_A**2, 0.5*v_B**2]
plt.figure(figsize=(7.5, 4.5))
plt.bar(labels, Ep_por_massa, label="Potencial por unidade de massa")
plt.bar(labels, Ec_por_massa, bottom=Ep_por_massa, label="Cinética por unidade de massa")
plt.ylabel("Energia por unidade de massa (J/kg)")
plt.title("A energia potencial vira energia cinética")
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.show()
```

## 5. Como interpretar o resultado

Nos dois pontos, a energia mecânica por unidade de massa é $70\,\mathrm{J/kg}$. Em A ela é inteiramente potencial; em B, parte da energia potencial foi convertida em cinética.

## 6. Exercício de fixação

Um corpo passa do repouso em $h_A=5\,\mathrm{m}$ para $h_B=1\,\mathrm{m}$, sem atrito. Use $g=10\,\mathrm{m/s^2}$ e calcule sua velocidade em B.

Tente prever o resultado antes de executar a célula seguinte.

```python
h_A_ex, h_B_ex, v_A_ex, g_ex = 5.0, 1.0, 0.0, 10.0
v_B_ex = (v_A_ex**2 + 2*g_ex*(h_A_ex-h_B_ex))**0.5
print(f"Velocidade em B = {v_B_ex:.2f} m/s")
```

## 7. Resumo

- Sem atrito: $mgh_A+\frac12mv_A^2=mgh_B+\frac12mv_B^2$.
- A massa pode ser cancelada.
- A energia potencial que diminui reaparece como energia cinética no modelo ideal.
