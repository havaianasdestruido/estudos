# Energia Potencial Elástica
### Trabalho, energia e suas transformações

**Objetivo:** Entender e calcular a energia armazenada ao comprimir ou distender uma mola ideal.

### Como usar este notebook

Execute as células na ordem, de cima para baixo. As células Markdown apresentam os conceitos; as células de código fazem as contas e geram gráficos com **NumPy** e **Matplotlib**.

> **Convenção:** neste material, usamos principalmente $W$ para trabalho, medido em joules (J). Algumas fontes usam $\tau$, mas esse símbolo também é muito utilizado para torque. O trabalho é uma grandeza escalar.

## 1. Conceito principal

Uma mola ideal armazena energia quando é comprimida ou esticada em relação à posição de equilíbrio. Pela lei de Hooke, a força elástica é proporcional à deformação; a energia armazenada é:

$$
\boxed{E_{pe}=\frac{kx^2}{2}}
$$

Aqui, $k$ é a constante elástica, em $\mathrm{N/m}$, e $x$ é o módulo da deformação, em metros. Como a deformação aparece ao quadrado, compressão e extensão de mesmo módulo armazenam a mesma energia no modelo ideal.

## 2. Exemplo resolvido passo a passo

Considere uma mola de constante $k=200\,\mathrm{N/m}$ comprimida em $x=0{,}10\,\mathrm{m}$. São valores ilustrativos, já que a tabela não fornece números.

$$
E_{pe}=\frac{200\cdot(0{,}10)^2}{2}=\boxed{1\,\mathrm{J}}
$$

Ao soltar a mola, essa energia pode transformar-se em energia cinética do corpo acoplado.

## 3. Conferindo o cálculo com Python

A próxima célula implementa a fórmula usando as grandezas e unidades indicadas no exemplo.

```python
k = 200.0   # N/m; exemplo ilustrativo
x = 0.10    # m
Epe = 0.5 * k * x**2
print(f"Energia potencial elástica = {Epe:.3f} J")
```

## 4. Visualização

O gráfico abaixo ajuda a interpretar a relação física, além de conferir os valores numéricos.

```python
import matplotlib.pyplot as plt
import numpy as np

x_vals = np.linspace(-0.20, 0.20, 201)
E_vals = 0.5 * k * x_vals**2
plt.figure(figsize=(8, 4.5))
plt.plot(x_vals, E_vals)
plt.scatter([x], [Epe], zorder=3, label=f"x = {x:.2f} m → {Epe:.2f} J")
plt.axvline(0, linewidth=0.8)
plt.xlabel("Deformação x (m; negativa = compressão neste esquema)")
plt.ylabel("Energia potencial elástica (J)")
plt.title("A energia elástica depende do quadrado da deformação")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## 5. Como interpretar o resultado

A curva é simétrica em relação a $x=0$. Dobrar o módulo da deformação quadruplica a energia armazenada. Em $x=0$, a mola ideal não está deformada e sua energia potencial elástica é zero.

## 6. Exercício de fixação

Uma mola com $k=100\,\mathrm{N/m}$ é esticada em $0{,}20\,\mathrm{m}$. Quanta energia elástica ela armazena?

Tente prever o resultado antes de executar a célula seguinte.

```python
k_ex, x_ex = 100.0, 0.20
Epe_ex = 0.5 * k_ex * x_ex**2
print(f"Epe = {Epe_ex:.2f} J")
```

## 7. Resumo

- $E_{pe}=\frac12 kx^2$.
- A deformação deve ser expressa em metros.
- Para mola ideal, dobrar $|x|$ quadruplica a energia potencial elástica.
