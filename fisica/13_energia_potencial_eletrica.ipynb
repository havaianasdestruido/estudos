# Energia Potencial Elétrica: Atração e Repulsão
### Trabalho, energia e suas transformações

**Objetivo:** Relacionar a energia potencial elétrica ao sinal das cargas e à distância entre elas.

### Como usar este notebook

Execute as células na ordem, de cima para baixo. As células Markdown apresentam os conceitos; as células de código fazem as contas e geram gráficos com **NumPy** e **Matplotlib**.

> **Convenção:** neste material, usamos principalmente $W$ para trabalho, medido em joules (J). Algumas fontes usam $\tau$, mas esse símbolo também é muito utilizado para torque. O trabalho é uma grandeza escalar.

## 1. Conceito principal

Duas cargas elétricas puntiformes interagem por meio da força elétrica. Escolhendo a energia potencial como zero quando a distância tende ao infinito, a energia potencial elétrica do par é:

$$
\boxed{E_{pe}=k_e\frac{q_1q_2}{r}}
$$

$k_e\approx8{,}99\times10^9\,\mathrm{N\,m^2/C^2}$, as cargas $q_1,q_2$ devem estar em coulombs e $r$ em metros. Cargas de mesmo sinal dão energia potencial positiva; cargas de sinais opostos, negativa. Em um sistema isolado, a diminuição da energia potencial pode virar aumento da energia cinética.

## 2. Exemplo resolvido passo a passo

Considere duas cargas positivas $q_1=+2\,\mu\mathrm{C}$ e $q_2=+3\,\mu\mathrm{C}$, separadas por $r_i=0{,}30\,\mathrm{m}$. Se a distância aumenta para $r_f=0{,}60\,\mathrm{m}$, a energia potencial inicial e final são:

$$
U_i=k_e\frac{q_1q_2}{r_i}\approx0{,}180\,\mathrm{J}
$$
$$
U_f=k_e\frac{q_1q_2}{r_f}\approx0{,}090\,\mathrm{J}
$$

Logo, $\Delta U=U_f-U_i\approx-0{,}090\,\mathrm{J}$. Se não houver outras transferências, a energia cinética aumenta em aproximadamente $0{,}090\,\mathrm{J}$.

## 3. Conferindo o cálculo com Python

A próxima célula implementa a fórmula usando as grandezas e unidades indicadas no exemplo.

```python
ke = 8.99e9
q1 = 2e-6       # C
q2 = 3e-6       # C
r_i = 0.30      # m
r_f = 0.60      # m
U_i = ke * q1 * q2 / r_i
U_f = ke * q1 * q2 / r_f
delta_U = U_f - U_i

print(f"Energia potencial inicial: {U_i:.3f} J")
print(f"Energia potencial final:   {U_f:.3f} J")
print(f"Variação da energia potencial: {delta_U:+.3f} J")
print(f"Aumento de energia cinética idealizado: {-delta_U:.3f} J")
```

## 4. Visualização

O gráfico abaixo ajuda a interpretar a relação física, além de conferir os valores numéricos.

```python
import matplotlib.pyplot as plt
import numpy as np

r_vals = np.linspace(0.15, 1.0, 300)
U_vals = ke * q1 * q2 / r_vals
plt.figure(figsize=(8, 4.5))
plt.plot(r_vals, U_vals)
plt.scatter([r_i, r_f], [U_i, U_f], zorder=3, label="Posições inicial e final")
plt.xlabel("Distância entre as cargas r (m)")
plt.ylabel("Energia potencial elétrica (J)")
plt.title("Duas cargas positivas: U diminui quando r aumenta")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## 5. Como interpretar o resultado

Como as duas cargas têm o mesmo sinal, elas se repelem. Ao afastá-las, a energia potencial diminui em direção a zero. Para cargas de sinais opostos, a energia potencial seria negativa sob essa convenção; o comportamento exato depende de como as cargas se movem e de sua energia total.

## 6. Exercício de fixação

Duas cargas têm $q_1=+1\,\mu\mathrm{C}$ e $q_2=-2\,\mu\mathrm{C}$, separadas por $0{,}20\,\mathrm{m}$. Calcule a energia potencial elétrica, usando o mesmo zero no infinito.

Tente prever o resultado antes de executar a célula seguinte.

```python
q1_ex, q2_ex, r_ex = 1e-6, -2e-6, 0.20
U_ex = ke * q1_ex * q2_ex / r_ex
print(f"Energia potencial elétrica = {U_ex:.3f} J")
```

## 7. Resumo

- $U=k_e q_1q_2/r$, com zero no infinito.
- Sinais iguais: $U>0$; sinais opostos: $U<0$.
- Em um sistema isolado, a energia mecânica total se conserva; energia potencial pode ser convertida em cinética.
