# Teorema do Trabalho e Energia Cinética
### Trabalho, energia e suas transformações

**Objetivo:** Relacionar o trabalho resultante realizado sobre um corpo à variação de sua energia cinética.

### Como usar este notebook

Execute as células na ordem, de cima para baixo. As células Markdown apresentam os conceitos; as células de código fazem as contas e geram gráficos com **NumPy** e **Matplotlib**.

> **Convenção:** neste material, usamos principalmente $W$ para trabalho, medido em joules (J). Algumas fontes usam $\tau$, mas esse símbolo também é muito utilizado para torque. O trabalho é uma grandeza escalar.

## 1. Conceito principal

O **teorema do trabalho e energia cinética** estabelece que o trabalho resultante de todas as forças sobre um corpo é igual à variação de sua energia cinética:

$$
\boxed{W_{\mathrm{resultante}}=\Delta E_c=E_{cf}-E_{ci}}
$$

A energia cinética é a energia associada ao movimento. Trabalho positivo aumenta a energia cinética; trabalho negativo a diminui. A conclusão se refere ao **trabalho resultante**, não necessariamente ao trabalho de uma única força.

## 2. Exemplo resolvido passo a passo

Um garoto chuta uma bola inicialmente parada. Considere que a bola passa a ter $5\,\mathrm{J}$ de energia cinética. Determine o trabalho resultante realizado sobre ela.

Dados: $E_{ci}=0\,\mathrm{J}$ e $E_{cf}=5\,\mathrm{J}$.

$$
W_{\mathrm{res}}=E_{cf}-E_{ci}=5-0=\boxed{+5\,\mathrm{J}}
$$

O sinal positivo indica ganho de energia cinética.

## 3. Conferindo o cálculo com Python

A próxima célula implementa a fórmula usando as grandezas e unidades indicadas no exemplo.

```python
E_ci = 0.0  # J
E_cf = 5.0  # J
W_resultante = E_cf - E_ci

print(f"Variação da energia cinética: ΔEc = {W_resultante:+.1f} J")
print(f"Trabalho resultante: W = {W_resultante:+.1f} J")
```

## 4. Visualização

O gráfico abaixo ajuda a interpretar a relação física, além de conferir os valores numéricos.

```python
import matplotlib.pyplot as plt

categorias = ["Inicial", "Final"]
energias = [E_ci, E_cf]
plt.figure(figsize=(7, 4.5))
plt.bar(categorias, energias)
plt.ylabel("Energia cinética (J)")
plt.title("A energia cinética aumenta após o chute")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

## 5. Como interpretar o resultado

O gráfico mostra que a energia cinética passou de $0$ para $5\,\mathrm{J}$. Portanto, o trabalho resultante foi positivo. Na situação idealizada, se outras forças realizarem trabalho desprezível, esse valor também corresponde ao trabalho do chute.

## 6. Exercício de fixação

Uma bola tem energia cinética inicial de $8\,\mathrm{J}$ e, depois de receber trabalho resultante, fica com $13\,\mathrm{J}$. Qual foi o trabalho resultante?

Tente prever o resultado antes de executar a célula seguinte.

```python
E_ci_ex = 8.0
E_cf_ex = 13.0
W_ex = E_cf_ex - E_ci_ex
print(f"W_resultante = {W_ex:+.1f} J")
```

## 7. Resumo

- Fórmula: $W_{\mathrm{res}}=\Delta E_c=E_{cf}-E_{ci}$.
- $W>0$: a energia cinética aumenta.
- $W<0$: a energia cinética diminui.
- O teorema usa o trabalho **resultante de todas as forças**.
