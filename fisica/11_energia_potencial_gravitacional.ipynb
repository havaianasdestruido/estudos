# Energia Potencial Gravitacional
### Trabalho, energia e suas transformações

**Objetivo:** Calcular a energia associada à posição de um corpo em um campo gravitacional próximo à superfície da Terra.

### Como usar este notebook

Execute as células na ordem, de cima para baixo. As células Markdown apresentam os conceitos; as células de código fazem as contas e geram gráficos com **NumPy** e **Matplotlib**.

> **Convenção:** neste material, usamos principalmente $W$ para trabalho, medido em joules (J). Algumas fontes usam $\tau$, mas esse símbolo também é muito utilizado para torque. O trabalho é uma grandeza escalar.

## 1. Conceito principal

Próximo à superfície da Terra, a energia potencial gravitacional em relação a um nível de referência pode ser calculada por:

$$
\boxed{E_p=mgh}
$$

- $m$: massa, em quilogramas;
- $g$: aceleração da gravidade, em $\mathrm{m/s^2}$;
- $h$: altura relativa ao nível escolhido, em metros.

O valor depende do nível de referência adotado. A variação de energia potencial entre dois níveis não depende da escolha do zero.

## 2. Exemplo resolvido passo a passo

Um objeto de $2\,\mathrm{kg}$ está a $4\,\mathrm{m}$ de altura em relação ao chão. Adote $g=10\,\mathrm{m/s^2}$:

$$
E_p=2\cdot10\cdot4=\boxed{80\,\mathrm{J}}
$$

Os valores são ilustrativos e o chão foi escolhido como nível de energia potencial zero.

## 3. Conferindo o cálculo com Python

A próxima célula implementa a fórmula usando as grandezas e unidades indicadas no exemplo.

```python
m = 2.0       # kg
g = 10.0      # m/s²; aproximação usada no exercício
h = 4.0       # m, acima do nível de referência
Ep = m * g * h
print(f"Energia potencial gravitacional = {Ep:.1f} J")
```

## 4. Visualização

O gráfico abaixo ajuda a interpretar a relação física, além de conferir os valores numéricos.

```python
import matplotlib.pyplot as plt
import numpy as np

alturas = np.linspace(0, 10, 101)
energias = m * g * alturas
plt.figure(figsize=(8, 4.5))
plt.plot(alturas, energias)
plt.scatter([h], [Ep], zorder=3, label=f"h = {h:.1f} m → {Ep:.1f} J")
plt.xlabel("Altura em relação ao referencial (m)")
plt.ylabel("Energia potencial gravitacional (J)")
plt.title("A energia potencial aumenta linearmente com a altura")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## 5. Como interpretar o resultado

Para massa e gravidade fixas, a energia potencial é diretamente proporcional à altura. Se a altura dobrar de $4$ para $8\,\mathrm{m}$, a energia potencial passa de $80$ para $160\,\mathrm{J}$, usando o mesmo referencial.

## 6. Exercício de fixação

Qual é a energia potencial de um objeto de $3\,\mathrm{kg}$, a $6\,\mathrm{m}$ do chão, usando $g=10\,\mathrm{m/s^2}$?

Tente prever o resultado antes de executar a célula seguinte.

```python
m_ex, h_ex, g_ex = 3.0, 6.0, 10.0
Ep_ex = m_ex * g_ex * h_ex
print(f"Ep = {Ep_ex:.1f} J")
```

## 7. Resumo

- $E_p=mgh$ perto da superfície terrestre.
- O resultado depende do nível de referência da altura.
- Ao cair, um corpo pode transformar energia potencial em cinética se outras perdas forem desprezíveis.
