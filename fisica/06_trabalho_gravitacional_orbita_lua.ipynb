# Trabalho da Gravidade no Movimento Orbital da Lua
### Trabalho, energia e suas transformações

**Objetivo:** Explicar por que a força gravitacional não realiza trabalho instantâneo em uma órbita circular ideal.

### Como usar este notebook

Execute as células na ordem, de cima para baixo. As células Markdown apresentam os conceitos; as células de código fazem as contas e geram gráficos com **NumPy** e **Matplotlib**.

> **Convenção:** neste material, usamos principalmente $W$ para trabalho, medido em joules (J). Algumas fontes usam $\tau$, mas esse símbolo também é muito utilizado para torque. O trabalho é uma grandeza escalar.

## 1. Conceito principal

Em uma órbita circular ideal, a força gravitacional da Terra aponta radialmente para o centro da órbita, enquanto a velocidade instantânea da Lua é tangente à trajetória. Assim, força e velocidade são perpendiculares.

No deslocamento infinitesimal, $dW=\vec F\cdot d\vec r=F\,ds\cos90^\circ=0$. Portanto, a gravidade não altera o módulo da velocidade em uma órbita perfeitamente circular: ela muda a **direção** do vetor velocidade.

A frase “a força gravitacional é perpendicular à velocidade” descreve a órbita circular ideal. Em uma órbita elíptica, essa perpendicularidade não vale em todos os pontos.

## 2. Exemplo resolvido passo a passo

Para uma órbita circular ideal, use $\theta=90^\circ$:

$$
W=F\,d\cos90^\circ=\boxed{0\,\mathrm{J}}
$$

A conclusão acima considera o trabalho ao longo do movimento circular ideal. A força continua sendo essencial: ela fornece a aceleração centrípeta necessária para manter a trajetória.

## 3. Conferindo o cálculo com Python

A próxima célula implementa a fórmula usando as grandezas e unidades indicadas no exemplo.

```python
import math

F_grav = 1.0  # N; módulo fictício apenas para ilustrar a geometria
ds = 100.0    # m; pequeno trecho ilustrativo
theta_deg = 90.0
W = F_grav * ds * math.cos(math.radians(theta_deg))
print(f"Trabalho no trecho perpendicular idealizado: {W:.2f} J")
print("A força pode alterar a direção da velocidade sem alterar seu módulo.")
```

## 4. Visualização

O gráfico abaixo ajuda a interpretar a relação física, além de conferir os valores numéricos.

```python
import matplotlib.pyplot as plt
import numpy as np

ang = np.linspace(0, 2*np.pi, 300)
R = 1.0
x, y = R*np.cos(ang), R*np.sin(ang)
# Um ponto na órbita e vetores radial (gravidade) e tangencial (velocidade)
px, py = R, 0.0
plt.figure(figsize=(6.5, 6.0))
plt.plot(x, y, label="Órbita circular ideal")
plt.arrow(px, py, -0.55, 0, head_width=0.08, length_includes_head=True, label="")
plt.arrow(px, py, 0, 0.55, head_width=0.08, length_includes_head=True, label="")
plt.text(0.42, -0.16, "Gravidade (radial)")
plt.text(1.02, 0.42, "Velocidade (tangente)")
plt.scatter([0], [0], label="Terra")
plt.scatter([px], [py], label="Lua")
plt.axis("equal")
plt.xlabel("x (unidades ilustrativas)")
plt.ylabel("y (unidades ilustrativas)")
plt.title("Força gravitacional e velocidade são perpendiculares")
plt.legend(loc="lower left")
plt.grid(alpha=0.3)
plt.show()
```

## 5. Como interpretar o resultado

A gravidade não realiza trabalho líquido ao longo de uma órbita circular completa e ideal. Em cada ponto, a força centrípeta altera a direção do movimento. Em uma órbita real não perfeitamente circular, a energia cinética e a potencial podem variar ao longo da trajetória.

## 6. Exercício de fixação

Em uma órbita circular ideal, uma força de módulo $F=100\,\mathrm{N}$ permanece perpendicular à velocidade. Qual é a potência instantânea $P=\vec F\cdot\vec v$?

Tente prever o resultado antes de executar a célula seguinte.

```python
F_ex, v_ex, theta_ex = 100.0, 500.0, 90.0
P_ex = F_ex * v_ex * math.cos(math.radians(theta_ex))
print(f"Potência instantânea = {P_ex:.1f} W")
```

## 7. Resumo

- Órbita circular ideal: gravidade radial e velocidade tangencial.
- Como $\vec F\perp\vec v$, a potência instantânea $P=\vec F\cdot\vec v=0$.
- A gravidade muda a direção da velocidade, não seu módulo, no modelo circular ideal.
