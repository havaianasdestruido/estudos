# Trabalho de uma Força Constante
### Trabalho, energia e suas transformações

**Objetivo:** Aplicar a expressão geral do trabalho e interpretar como o ângulo muda o resultado.

### Como usar este notebook

Execute as células na ordem, de cima para baixo. As células Markdown apresentam os conceitos; as células de código fazem as contas e geram gráficos com **NumPy** e **Matplotlib**.

> **Convenção:** neste material, usamos principalmente $W$ para trabalho, medido em joules (J). Algumas fontes usam $\tau$, mas esse símbolo também é muito utilizado para torque. O trabalho é uma grandeza escalar.

## 1. Conceito principal

Quando uma força constante $\vec F$ age sobre um corpo que se desloca por $\vec d$, o trabalho é o produto escalar entre os vetores:

$$
\boxed{W=F\,d\cos\theta}
$$

O trabalho depende apenas da componente da força paralela ao deslocamento. Uma força no mesmo sentido realiza trabalho positivo; no sentido oposto, negativo; perpendicular ao deslocamento, trabalho nulo.

## 2. Exemplo resolvido passo a passo

Considere uma força de $8\,\mathrm{N}$, deslocamento de $5\,\mathrm{m}$ e ângulo de $60^\circ$:

$$
W=8\cdot5\cdot\cos60^\circ=40\cdot0{,}5=\boxed{20\,\mathrm{J}}
$$

O exemplo numérico é ilustrativo; a tabela original apresenta a fórmula geral, sem valores específicos.

## 3. Conferindo o cálculo com Python

A próxima célula implementa a fórmula usando as grandezas e unidades indicadas no exemplo.

```python
import math

F = 8.0       # N; exemplo ilustrativo
d = 5.0       # m
theta_deg = 60.0
W = F * d * math.cos(math.radians(theta_deg))
print(f"Componente da força no deslocamento: {F * math.cos(math.radians(theta_deg)):.1f} N")
print(f"Trabalho: {W:.1f} J")
```

## 4. Visualização

O gráfico abaixo ajuda a interpretar a relação física, além de conferir os valores numéricos.

```python
import matplotlib.pyplot as plt
import numpy as np

angulos = np.linspace(0, 180, 181)
W_ang = F * d * np.cos(np.deg2rad(angulos))
plt.figure(figsize=(8, 4.5))
plt.plot(angulos, W_ang)
plt.scatter([theta_deg], [W], zorder=3, label=f"Exemplo: θ = {theta_deg:.0f}°")
plt.axhline(0, linewidth=0.8)
plt.xlabel("Ângulo θ (graus)")
plt.ylabel("Trabalho (J)")
plt.title("Trabalho de uma força constante conforme o ângulo")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## 5. Como interpretar o resultado

A curva cruza zero em $90^\circ$. Para o mesmo módulo de força e deslocamento, o trabalho máximo positivo ocorre a $0^\circ$, e o mais negativo ocorre a $180^\circ$.

## 6. Exercício de fixação

Uma força de $10\,\mathrm{N}$ atua a $30^\circ$ da direção do deslocamento, que mede $4\,\mathrm{m}$. Calcule o trabalho.

Tente prever o resultado antes de executar a célula seguinte.

```python
F_ex, d_ex, theta_ex = 10.0, 4.0, 30.0
W_ex = F_ex * d_ex * math.cos(math.radians(theta_ex))
print(f"W = {W_ex:.2f} J")
```

## 7. Resumo

- Fórmula geral: $W=Fd\cos\theta$.
- O ângulo é medido entre força e deslocamento.
- Somente a componente paralela ao deslocamento contribui para o trabalho.
