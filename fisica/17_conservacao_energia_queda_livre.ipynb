## 1. Conceito principal

Desprezando a resistência do ar, a energia potencial gravitacional inicial transforma-se em energia cinética:

\[
E_{pi}=E_{cf}\quad\Rightarrow\quad mgh=\frac{mv^2}{2}
\]

A massa aparece nos dois lados e pode ser cancelada:

\[
\boxed{h=\frac{v^2}{2g}}\qquad\boxed{v=\sqrt{2gh}}
\]

A velocidade deve estar em \(\mathrm{m/s}\), a altura em metros e \(g\) em \(\mathrm{m/s^2}\).

## 2. Exemplo resolvido passo a passo

Um objeto chega ao chão a \(72\,\mathrm{km/h}\). Primeiro convertemos a velocidade:

\[
72\div3{,}6=20\,\mathrm{m/s}
\]

Com \(g=10\,\mathrm{m/s^2}\):

\[
h=\frac{20^2}{2\cdot10}=\frac{400}{20}=\boxed{20\,\mathrm{m}}
\]

O resultado pressupõe que o objeto foi solto do repouso e que a resistência do ar é desprezível.

## 3. Conferindo o cálculo com Python

A próxima célula implementa a fórmula usando as grandezas e unidades indicadas no exemplo.

```python
v_kmh = 72.0
v = v_kmh / 3.6  # km/h -> m/s
g = 10.0
h = v**2 / (2*g)
print(f"Velocidade convertida: {v:.1f} m/s")
print(f"Altura estimada: {h:.1f} m")
```

## 4. Visualização

O gráfico abaixo ajuda a interpretar a relação física, além de conferir os valores numéricos.

```python
import matplotlib.pyplot as plt
import numpy as np

alturas = np.linspace(0, 25, 101)
velocidades = np.sqrt(2*g*alturas)
plt.figure(figsize=(8, 4.5))
plt.plot(alturas, velocidades)
plt.scatter([h], [v], zorder=3, label=f"{h:.0f} m → {v:.0f} m/s")
plt.xlabel("Altura de queda (m)")
plt.ylabel("Velocidade de impacto (m/s)")
plt.title("Velocidade de queda em função da altura")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## 5. Como interpretar o resultado

A velocidade aumenta com a raiz quadrada da altura, enquanto a energia potencial é diretamente proporcional a ela. Dobrar a altura não dobra a velocidade; multiplica-a por \(\sqrt2\).

## 6. Exercício de fixação

Um objeto cai do repouso de uma altura de \(45\,\mathrm{m}\). Desprezando o ar, determine a velocidade final usando \(g=10\,\mathrm{m/s^2}\).

Tente prever o resultado antes de executar a célula seguinte.

```python
h_ex, g_ex = 45.0, 10.0
v_ex = (2*g_ex*h_ex)**0.5
print(f"Velocidade final = {v_ex:.2f} m/s")
```
