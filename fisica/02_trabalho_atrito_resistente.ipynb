# Trabalho da Força de Atrito: Trabalho Resistente
### Trabalho, energia e suas transformações

**Objetivo:** Entender por que o trabalho do atrito pode ser negativo quando um objeto desliza e para.

### Como usar este notebook

Execute as células na ordem, de cima para baixo. As células Markdown apresentam os conceitos; as células de código fazem as contas e geram gráficos com **NumPy** e **Matplotlib**.

> **Convenção:** neste material, usamos principalmente $W$ para trabalho, medido em joules (J). Algumas fontes usam $\tau$, mas esse símbolo também é muito utilizado para torque. O trabalho é uma grandeza escalar.

## 1. Conceito principal

O atrito cinético geralmente atua em sentido oposto ao movimento relativo entre as superfícies. Quando um objeto desliza sobre um chão e para devido ao atrito, sua energia cinética diminui.

Se o atrito for a única força que realiza trabalho significativo, o teorema trabalho–energia dá:

$$
\boxed{W_{\mathrm{atrito}}=\Delta E_c=E_{cf}-E_{ci}}
$$

O trabalho negativo indica que a energia cinética foi transferida para outras formas, como energia térmica.

## 2. Exemplo resolvido passo a passo

Um objeto desliza com energia cinética inicial de $5\,\mathrm{J}$ e para. Sua energia cinética final é zero.

$$
W_{\mathrm{atrito}}=0-5=\boxed{-5\,\mathrm{J}}
$$

A hipótese é que o atrito seja responsável pela redução da energia cinética; a energia não desaparece, ela é dissipada principalmente como energia interna.

## 3. Conferindo o cálculo com Python

A próxima célula implementa a fórmula usando as grandezas e unidades indicadas no exemplo.

```python
E_ci = 5.0  # J
E_cf = 0.0  # J
W_atrito = E_cf - E_ci

print(f"ΔEc = {E_cf:.1f} - {E_ci:.1f} = {W_atrito:.1f} J")
print("O sinal negativo indica trabalho resistente.")
```

## 4. Visualização

O gráfico abaixo ajuda a interpretar a relação física, além de conferir os valores numéricos.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 4.5))
plt.bar(["Inicial", "Final"], [E_ci, E_cf])
plt.ylabel("Energia cinética (J)")
plt.title("O objeto desliza até parar")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

## 5. Como interpretar o resultado

A energia cinética cai em $5\,\mathrm{J}$. Nesse modelo, o trabalho do atrito é $-5\,\mathrm{J}$. O sinal não representa energia “negativa” armazenada: representa transferência de energia para fora da energia cinética do objeto.

## 6. Exercício de fixação

Um carrinho perde velocidade: sua energia cinética passa de $12\,\mathrm{J}$ para $3\,\mathrm{J}$. Se o atrito for a única força que realiza trabalho, calcule seu trabalho.

Tente prever o resultado antes de executar a célula seguinte.

```python
Ec_i_ex, Ec_f_ex = 12.0, 3.0
W_atrito_ex = Ec_f_ex - Ec_i_ex
print(f"Trabalho do atrito: {W_atrito_ex:+.1f} J")
```

## 7. Resumo

- Para um corpo que para por ação do atrito: $W_{\mathrm{atrito}}=E_{cf}-E_{ci}<0$.
- Trabalho resistente diminui a energia cinética.
- A energia é transformada, em geral, em energia térmica e deformações microscópicas.
