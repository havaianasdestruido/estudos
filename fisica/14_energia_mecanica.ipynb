# Energia Mecânica
### Trabalho, energia e suas transformações

**Objetivo:** Calcular a energia mecânica como soma das energias cinética e potencial relevantes.

### Como usar este notebook

Execute as células na ordem, de cima para baixo. As células Markdown apresentam os conceitos; as células de código fazem as contas e geram gráficos com **NumPy** e **Matplotlib**.

> **Convenção:** neste material, usamos principalmente $W$ para trabalho, medido em joules (J). Algumas fontes usam $\tau$, mas esse símbolo também é muito utilizado para torque. O trabalho é uma grandeza escalar.

## 1. Conceito principal

A energia mecânica de um sistema é a soma da energia cinética com as energias potenciais consideradas:

$$
\boxed{E_m=E_c+E_p}
$$

Em problemas gravitacionais, muitas vezes $E_p=mgh$. Se houver mola, pode-se incluir $E_{pe}=\frac12kx^2$; se houver interações elétricas, inclui-se a energia potencial elétrica adequada. A energia mecânica só permanece constante quando não há trabalho líquido de forças não conservativas, como o atrito, no sistema analisado.

## 2. Exemplo resolvido passo a passo

Um objeto de $2\,\mathrm{kg}$ move-se a $3\,\mathrm{m/s}$ e está a $4\,\mathrm{m}$ acima do nível de referência. Com $g=10\,\mathrm{m/s^2}$:

$$
E_c=\frac12(2)(3^2)=9\,\mathrm{J}
$$
$$
E_p=(2)(10)(4)=80\,\mathrm{J}
$$
$$
E_m=9+80=\boxed{89\,\mathrm{J}}
$$

## 3. Conferindo o cálculo com Python

A próxima célula implementa a fórmula usando as grandezas e unidades indicadas no exemplo.

```python
m, v, g, h = 2.0, 3.0, 10.0, 4.0
Ec = 0.5 * m * v**2
Ep = m * g * h
Em = Ec + Ep
print(f"Energia cinética:           {Ec:.1f} J")
print(f"Energia potencial:          {Ep:.1f} J")
print(f"Energia mecânica (Ec + Ep): {Em:.1f} J")
```

## 4. Visualização

O gráfico abaixo ajuda a interpretar a relação física, além de conferir os valores numéricos.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7.5, 4.5))
plt.bar(["Cinética", "Potencial", "Mecânica total"], [Ec, Ep, Em])
plt.ylabel("Energia (J)")
plt.title("Composição da energia mecânica")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

## 5. Como interpretar o resultado

Neste exemplo, a energia potencial é maior que a cinética, mas ambas fazem parte da energia mecânica. O gráfico mostra os dois componentes e sua soma. Não some a energia mecânica novamente com seus componentes: $E_m$ já é o total.

## 6. Exercício de fixação

Um corpo tem $E_c=15\,\mathrm{J}$ e energia potencial de $25\,\mathrm{J}$. Calcule sua energia mecânica.

Tente prever o resultado antes de executar a célula seguinte.

```python
Ec_ex, Ep_ex = 15.0, 25.0
Em_ex = Ec_ex + Ep_ex
print(f"Energia mecânica = {Em_ex:.1f} J")
```
