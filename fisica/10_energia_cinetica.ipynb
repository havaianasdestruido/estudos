# Energia Cinética
### Trabalho, energia e suas transformações

**Objetivo:** Calcular a energia associada ao movimento e observar como ela depende da velocidade.

### Como usar este notebook

Execute as células na ordem, de cima para baixo. As células Markdown apresentam os conceitos; as células de código fazem as contas e geram gráficos com **NumPy** e **Matplotlib**.

> **Convenção:** neste material, usamos principalmente $W$ para trabalho, medido em joules (J). Algumas fontes usam $\tau$, mas esse símbolo também é muito utilizado para torque. O trabalho é uma grandeza escalar.

## 1. Conceito principal

A energia cinética translacional de um corpo de massa $m$, que se move com velocidade de módulo $v$, é:

$$
\boxed{E_c=\frac{mv^2}{2}}
$$

A massa aparece linearmente; a velocidade aparece ao quadrado. Assim, mantendo a massa fixa, dobrar a velocidade quadruplica a energia cinética. Se $v=0$, a energia cinética é zero.

## 2. Exemplo resolvido passo a passo

Use um corpo de massa $m=2\,\mathrm{kg}$ que se move a $v=3\,\mathrm{m/s}$ (valores ilustrativos):

$$
E_c=\frac{2\cdot3^2}{2}=\boxed{9\,\mathrm{J}}
$$

A energia cinética nunca é negativa, pois a massa é positiva e a velocidade aparece elevada ao quadrado.

## 3. Conferindo o cálculo com Python

A próxima célula implementa a fórmula usando as grandezas e unidades indicadas no exemplo.

```python
m = 2.0  # kg; exemplo ilustrativo
v = 3.0  # m/s
Ec = 0.5 * m * v**2
print(f"Ec = ½ · {m:.1f} · {v:.1f}² = {Ec:.1f} J")
```

## 4. Visualização

O gráfico abaixo ajuda a interpretar a relação física, além de conferir os valores numéricos.

```python
import matplotlib.pyplot as plt
import numpy as np

velocidades = np.linspace(0, 10, 101)
energias = 0.5 * m * velocidades**2
plt.figure(figsize=(8, 4.5))
plt.plot(velocidades, energias)
plt.scatter([v], [Ec], zorder=3, label=f"Exemplo: {v:.1f} m/s → {Ec:.1f} J")
plt.xlabel("Velocidade (m/s)")
plt.ylabel("Energia cinética (J)")
plt.title("A energia cinética cresce com o quadrado da velocidade")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## 5. Como interpretar o resultado

O gráfico é uma parábola: o crescimento não é linear. Por exemplo, uma velocidade de $6\,\mathrm{m/s}$, duas vezes a velocidade usada no exemplo, corresponde a quatro vezes a energia cinética: $36\,\mathrm{J}$.

## 6. Exercício de fixação

Calcule a energia cinética de uma bicicleta e ciclista de massa total $80\,\mathrm{kg}$, movendo-se a $5\,\mathrm{m/s}$.

Tente prever o resultado antes de executar a célula seguinte.

```python
m_ex, v_ex = 80.0, 5.0
Ec_ex = 0.5 * m_ex * v_ex**2
print(f"Energia cinética = {Ec_ex:.1f} J")
```

- $E_c=\frac12 mv^2$.
- Dobrar a massa dobra $E_c$; dobrar a velocidade quadruplica $E_c$.
