# Trabalho da Força de Atrito pela Fórmula Geral
### Trabalho, energia e suas transformações

**Objetivo:** Calcular o trabalho de uma força de atrito constante por meio do ângulo entre força e deslocamento.

### Como usar este notebook

Execute as células na ordem, de cima para baixo. As células Markdown apresentam os conceitos; as células de código fazem as contas e geram gráficos com **NumPy** e **Matplotlib**.

> **Convenção:** neste material, usamos principalmente $W$ para trabalho, medido em joules (J). Algumas fontes usam $\tau$, mas esse símbolo também é muito utilizado para torque. O trabalho é uma grandeza escalar.

## 1. Conceito principal

Para uma força constante, o trabalho é calculado por:

$$
\boxed{W=F\,d\cos\theta}
$$

Aqui, $F$ é o módulo da força, $d$ é o módulo do deslocamento e $\theta$ é o ângulo entre o vetor força e o vetor deslocamento. Para o atrito em oposição ao movimento, $\theta=180^\circ$ e $\cos 180^\circ=-1$.

## 2. Exemplo resolvido passo a passo

Uma força de atrito de módulo $3\,\mathrm{N}$ atua contra o deslocamento de $10\,\mathrm{m}$.

$$
W_{\mathrm{at}}=3\cdot10\cdot\cos(180^\circ)
=30(-1)=\boxed{-30\,\mathrm{J}}
$$

O trabalho é negativo porque a força aponta na direção oposta ao deslocamento.

## 3. Conferindo o cálculo com Python

A próxima célula implementa a fórmula usando as grandezas e unidades indicadas no exemplo.

```python
import math

F_at = 3.0       # N
d = 10.0          # m
theta_deg = 180.0
W_at = F_at * d * math.cos(math.radians(theta_deg))
print(f"cos(θ) = {math.cos(math.radians(theta_deg)):.1f}")
print(f"W_atrito = {W_at:.1f} J")
```

## 4. Visualização

O gráfico abaixo ajuda a interpretar a relação física, além de conferir os valores numéricos.

```python
import matplotlib.pyplot as plt
import numpy as np

angulos = np.linspace(0, 180, 181)
trabalhos = F_at * d * np.cos(np.deg2rad(angulos))
plt.figure(figsize=(8, 4.5))
plt.plot(angulos, trabalhos)
plt.scatter([theta_deg], [W_at], zorder=3, label=f"180° → {W_at:.0f} J")
plt.axhline(0, linewidth=0.8)
plt.xlabel("Ângulo entre a força e o deslocamento (graus)")
plt.ylabel("Trabalho (J)")
plt.title("O trabalho depende do cosseno do ângulo")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## 5. Como interpretar o resultado

Quando o ângulo é $180^\circ$, o cosseno vale $-1$, e o trabalho é o mais negativo possível para os módulos fixos de força e deslocamento. Se a força fosse perpendicular, o trabalho seria zero.

## 6. Exercício de fixação

Uma força de atrito constante de $4\,\mathrm{N}$ atua contra um deslocamento de $6\,\mathrm{m}$. Determine o trabalho.

Tente prever o resultado antes de executar a célula seguinte.

```python
F_ex, d_ex, theta_ex = 4.0, 6.0, 180.0
W_ex = F_ex * d_ex * math.cos(math.radians(theta_ex))
print(f"W = {W_ex:.1f} J")
```

## 7. Resumo

- $W=Fd\cos\theta$.
- Atrito oposto ao deslocamento: $\theta=180^\circ$, então $W=-Fd$.
- A unidade de trabalho é o joule: $1\,\mathrm{J}=1\,\mathrm{N\,m}$.
