# Trabalho Resultante e Variação da Energia Cinética
### Trabalho, energia e suas transformações

**Objetivo:** Somar o trabalho de várias forças e relacionar o resultado à variação da energia cinética.

### Como usar este notebook

Execute as células na ordem, de cima para baixo. As células Markdown apresentam os conceitos; as células de código fazem as contas e geram gráficos com **NumPy** e **Matplotlib**.

> **Convenção:** neste material, usamos principalmente $W$ para trabalho, medido em joules (J). Algumas fontes usam $\tau$, mas esse símbolo também é muito utilizado para torque. O trabalho é uma grandeza escalar.

## 1. Conceito principal

O trabalho resultante é a soma algébrica dos trabalhos individuais:

$$
W_{\mathrm{total}}=W_F+W_{\mathrm{atrito}}+W_P+W_N
$$

Pelo teorema trabalho–energia:

$$
\boxed{W_{\mathrm{total}}=\Delta E_c}
$$

Os sinais importam: trabalhos motores são positivos e trabalhos resistentes são negativos. Trabalho resultante positivo significa aumento da energia cinética, não necessariamente aumento de toda forma de energia do sistema.

## 2. Exemplo resolvido passo a passo

Para os trabalhos $W_F=40\,\mathrm{J}$, $W_{\mathrm{atrito}}=-30\,\mathrm{J}$, $W_P=0$ e $W_N=0$:

$$
W_{\mathrm{total}}=40-30+0+0=\boxed{+10\,\mathrm{J}}
$$

Logo, a energia cinética aumenta em $10\,\mathrm{J}$.

## 3. Conferindo o cálculo com Python

A próxima célula implementa a fórmula usando as grandezas e unidades indicadas no exemplo.

```python
W_F = 40.0
W_atrito = -30.0
W_P = 0.0
W_N = 0.0
W_total = W_F + W_atrito + W_P + W_N

print(f"Trabalho da força aplicada: {W_F:+.1f} J")
print(f"Trabalho do atrito:         {W_atrito:+.1f} J")
print(f"Trabalho do peso:           {W_P:+.1f} J")
print(f"Trabalho da normal:         {W_N:+.1f} J")
print(f"Trabalho resultante:        {W_total:+.1f} J")
print(f"Variação da energia cinética: ΔEc = {W_total:+.1f} J")
```

## 4. Visualização

O gráfico abaixo ajuda a interpretar a relação física, além de conferir os valores numéricos.

```python
import matplotlib.pyplot as plt

nomes = ["Força aplicada", "Atrito", "Peso", "Normal", "Resultante"]
valores = [W_F, W_atrito, W_P, W_N, W_total]
plt.figure(figsize=(9, 4.8))
plt.bar(nomes, valores)
plt.axhline(0, linewidth=0.8)
plt.ylabel("Trabalho (J)")
plt.title("Contribuição de cada força para o trabalho total")
plt.xticks(rotation=15, ha="right")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

## 5. Como interpretar o resultado

A força aplicada fornece $40\,\mathrm{J}$, enquanto o atrito retira $30\,\mathrm{J}$ da energia cinética. O saldo é $+10\,\mathrm{J}$. Para converter esse valor em uma velocidade final, seria necessário conhecer a energia cinética inicial ou a massa e velocidade inicial.

## 6. Exercício de fixação

Uma força realiza $+25\,\mathrm{J}$ de trabalho e o atrito realiza $-9\,\mathrm{J}$. Peso e normal realizam trabalho nulo. Qual é a variação da energia cinética?

Tente prever o resultado antes de executar a célula seguinte.

```python
W_ex = 25.0 - 9.0 + 0.0 + 0.0
print(f"ΔEc = {W_ex:+.1f} J")
```
