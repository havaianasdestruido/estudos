# Trabalho de uma Força Inclinada
### Trabalho, energia e suas transformações

**Objetivo:** Usar a componente da força paralela ao deslocamento para calcular o trabalho.

### Como usar este notebook

Execute as células na ordem, de cima para baixo. As células Markdown apresentam os conceitos; as células de código fazem as contas e geram gráficos com **NumPy** e **Matplotlib**.

> **Convenção:** neste material, usamos principalmente \(W\) para trabalho, medido em joules (J). Algumas fontes usam \(\tau\), mas esse símbolo também é muito utilizado para torque. O trabalho é uma grandeza escalar.

## 1. Conceito principal

Quando uma força é inclinada, só a componente paralela ao deslocamento contribui para o trabalho. Para uma força \(F\) inclinada de um ângulo \(\alpha\) em relação ao deslocamento horizontal:

\[
F_x=F\cos\alpha,\qquad W_F=F\,d\cos\alpha=F_xd
\]

Se já conhecemos a componente horizontal, não é necessário conhecer separadamente o módulo total da força e seu ângulo.

## 2. Exemplo resolvido passo a passo

A componente horizontal da força é \(F_x=4\,\mathrm{N}\), e o bloco se desloca \(10\,\mathrm{m}\):

\[
W_F=F_xd=4\cdot10=\boxed{+40\,\mathrm{J}}
\]

O trabalho é positivo se a componente horizontal aponta no mesmo sentido do deslocamento.

## 3. Conferindo o cálculo com Python

A próxima célula implementa a fórmula usando as grandezas e unidades indicadas no exemplo.

```python
F_x = 4.0  # N, componente ao longo do deslocamento
d = 10.0   # m
W_F = F_x * d
print(f"W_F = F_x · d = {F_x:.1f} · {d:.1f} = {W_F:.1f} J")
```

## 4. Visualização

O gráfico abaixo ajuda a interpretar a relação física, além de conferir os valores numéricos.

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.plot([0, d], [0, 0], linewidth=3, label="Deslocamento")
ax.arrow(2, 0, 2.0, 1.8, head_width=0.12, length_includes_head=True)
ax.arrow(2, 0, 2.0, 0.0, head_width=0.12, length_includes_head=True)
ax.text(4.1, 1.65, "Força inclinada F")
ax.text(4.1, -0.3, "Componente Fₓ")
ax.text(6, 0.25, "d")
ax.set_xlim(0, 10)
ax.set_ylim(-0.7, 2.5)
ax.set_yticks([])
ax.set_xlabel("Direção do deslocamento (m; esquema, não em escala)")
ax.set_title("Apenas a componente paralela realiza trabalho")
ax.legend(loc="upper right")
ax.grid(axis="x", alpha=0.3)
plt.show()
```

## 5. Como interpretar o resultado

O valor de \(40\,\mathrm{J}\) é calculado com a componente paralela, não com o módulo total da força. A parte vertical da força não realiza trabalho durante um deslocamento estritamente horizontal.

## 6. Exercício de fixação

Uma força inclinada tem componente horizontal de \(7\,\mathrm{N}\) e desloca um bloco \(3\,\mathrm{m}\) no mesmo sentido. Determine o trabalho dessa força.

Tente prever o resultado antes de executar a célula seguinte.

```python
Fx_ex, d_ex = 7.0, 3.0
W_ex = Fx_ex * d_ex
print(f"Trabalho da força = {W_ex:.1f} J")
```

## 7. Resumo

- \(F_x=F\cos\alpha\).
- \(W=F_xd=Fd\cos\alpha\).
