# Lançamento Oblíquo e Conservação de Energia
### Trabalho, energia e suas transformações

**Objetivo:** Encontrar a velocidade no ponto mais alto de um lançamento oblíquo por meio da energia mecânica.

### Como usar este notebook

Execute as células na ordem, de cima para baixo. As células Markdown apresentam os conceitos; as células de código fazem as contas e geram gráficos com **NumPy** e **Matplotlib**.

> **Convenção:** neste material, usamos principalmente $W$ para trabalho, medido em joules (J). Algumas fontes usam $\tau$, mas esse símbolo também é muito utilizado para torque. O trabalho é uma grandeza escalar.

## 1. Conceito principal

Sem resistência do ar, a energia mecânica se conserva:

$$
E_m=E_p+E_c,\qquad E_c=\frac{mv^2}{2}
$$

No ponto mais alto de um lançamento oblíquo, a componente vertical da velocidade é zero, mas a componente horizontal permanece diferente de zero. Portanto, a velocidade total **não** é necessariamente zero.

## 2. Exemplo resolvido passo a passo

Um corpo de massa $1\,\mathrm{kg}$ é lançado do solo com energia mecânica $E_m=20\,\mathrm{J}$. No ponto mais alto, $E_p=7{,}5\,\mathrm{J}$.

$$
E_c=E_m-E_p=20-7{,}5=12{,}5\,\mathrm{J}
$$
$$
12{,}5=\frac{1\cdot v^2}{2}\Rightarrow v^2=25\Rightarrow\boxed{v=5\,\mathrm{m/s}}
$$

No topo, essa velocidade é horizontal no modelo ideal.

## 3. Conferindo o cálculo com Python

A próxima célula implementa a fórmula usando as grandezas e unidades indicadas no exemplo.

```python
m = 1.0       # kg
Em = 20.0     # J
Ep_topo = 7.5 # J
Ec_topo = Em - Ep_topo
v_topo = (2*Ec_topo/m)**0.5
h_topo = Ep_topo / (m*10.0)  # g = 10 m/s²
print(f"Energia cinética no topo: {Ec_topo:.1f} J")
print(f"Velocidade no topo: {v_topo:.1f} m/s")
print(f"Altura máxima: {h_topo:.2f} m")
```

## 4. Visualização

O gráfico abaixo ajuda a interpretar a relação física, além de conferir os valores numéricos.

```python
import matplotlib.pyplot as plt
import numpy as np

# Reconstrução idealizada da trajetória a partir de v_x = 5 m/s e da altura máxima.
g = 10.0
vy0 = (2*g*h_topo)**0.5
vx0 = v_topo
t_voo = 2*vy0/g
t = np.linspace(0, t_voo, 200)
x = vx0*t
y = vy0*t - 0.5*g*t**2
plt.figure(figsize=(8, 4.5))
plt.plot(x, y)
plt.scatter([vx0*(vy0/g)], [h_topo], zorder=3, label="Ponto mais alto")
plt.xlabel("Deslocamento horizontal (m)")
plt.ylabel("Altura (m)")
plt.title("Trajetória ideal de um lançamento oblíquo")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## 5. Como interpretar o resultado

A energia cinética no ponto mais alto é positiva, logo a velocidade horizontal também é positiva. A partir da altura máxima, podemos obter $v_{y0}=\sqrt{2gh_{\max}}$, enquanto $v_x=5\,\mathrm{m/s}$ permanece constante se não houver resistência do ar.

## 6. Exercício de fixação

Um projétil de massa $2\,\mathrm{kg}$ tem energia mecânica de $50\,\mathrm{J}$. Em seu ponto mais alto, sua energia potencial é $30\,\mathrm{J}$. Qual é sua velocidade nesse ponto?

Tente prever o resultado antes de executar a célula seguinte.

```python
m_ex, Em_ex, Ep_ex = 2.0, 50.0, 30.0
Ec_ex = Em_ex - Ep_ex
v_ex = (2*Ec_ex/m_ex)**0.5
print(f"Ec no topo = {Ec_ex:.1f} J")
print(f"Velocidade no topo = {v_ex:.2f} m/s")
```

## 7. Resumo

- $E_m=E_p+E_c$ permanece constante sem resistência do ar.
- No ponto mais alto, a componente vertical é zero.
- Em um lançamento oblíquo, a componente horizontal normalmente mantém a velocidade diferente de zero.
