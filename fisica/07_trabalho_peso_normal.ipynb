# Trabalho da Força Peso e da Força Normal
### Trabalho, energia e suas transformações

**Objetivo:** Calcular o trabalho do peso e da normal em um deslocamento horizontal.

### Como usar este notebook

Execute as células na ordem, de cima para baixo. As células Markdown apresentam os conceitos; as células de código fazem as contas e geram gráficos com **NumPy** e **Matplotlib**.

> **Convenção:** neste material, usamos principalmente $W$ para trabalho, medido em joules (J). Algumas fontes usam $\tau$, mas esse símbolo também é muito utilizado para torque. O trabalho é uma grandeza escalar.

## 1. Conceito principal

Em um deslocamento horizontal, o peso aponta verticalmente para baixo e a normal de uma superfície horizontal aponta verticalmente para cima. As duas forças fazem $90^\circ$ com o deslocamento.

$$
W_P=P\,d\cos90^\circ=0,\qquad W_N=N\,d\cos90^\circ=0
$$

Isso pressupõe uma superfície horizontal e ausência de deslocamento vertical.

## 2. Exemplo resolvido passo a passo

Um bloco percorre $10\,\mathrm{m}$ horizontalmente. Como o peso e a normal são perpendiculares ao deslocamento:

$$
\boxed{W_P=0\,\mathrm{J}},\qquad \boxed{W_N=0\,\mathrm{J}}
$$

O resultado não depende dos valores do peso ou da normal, contanto que a geometria descrita permaneça válida.

## 3. Conferindo o cálculo com Python

A próxima célula implementa a fórmula usando as grandezas e unidades indicadas no exemplo.

```python
import math

P = 50.0   # N; valor ilustrativo
N = 50.0   # N; plano horizontal sem aceleração vertical
s = 10.0   # m
theta = 90.0
W_P = P * s * math.cos(math.radians(theta))
W_N = N * s * math.cos(math.radians(theta))
print(f"Trabalho do peso: W_P = {W_P:.1f} J")
print(f"Trabalho da normal: W_N = {W_N:.1f} J")
```

## 4. Visualização

O gráfico abaixo ajuda a interpretar a relação física, além de conferir os valores numéricos.

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.plot([0, 10], [0, 0], linewidth=3, label="Deslocamento horizontal")
ax.arrow(4, 0, 0, -1.5, head_width=0.18, length_includes_head=True)
ax.arrow(4, 0, 0, 1.5, head_width=0.18, length_includes_head=True)
ax.text(4.2, -1.1, "Peso")
ax.text(4.2, 1.0, "Normal")
ax.text(6, 0.15, "Deslocamento")
ax.set_xlim(0, 10)
ax.set_ylim(-2, 2)
ax.set_xlabel("Posição horizontal (m; esquema)")
ax.set_yticks([])
ax.set_title("Peso e normal não têm componente horizontal")
ax.legend(loc="upper right")
ax.grid(axis="x", alpha=0.3)
plt.show()
```

## 5. Como interpretar o resultado

O trabalho de uma força depende da componente na direção do deslocamento. A normal e o peso podem equilibrar-se verticalmente, mas o motivo do trabalho nulo é que cada força é perpendicular ao deslocamento horizontal.

## 6. Exercício de fixação

O mesmo bloco se desloca $4\,\mathrm{m}$ horizontalmente. O trabalho do peso muda em relação ao exemplo?

Tente prever o resultado antes de executar a célula seguinte.

```python
d_ex = 4.0
P_ex = 50.0
theta_ex = 90.0
W_ex = P_ex * d_ex * math.cos(math.radians(theta_ex))
print(f"W_peso = {W_ex:.1f} J — continua sendo zero.")
```

## 7. Resumo

- Plano horizontal + deslocamento horizontal: $W_P=W_N=0$.
- A normal pode ter módulo diferente do peso em outras situações, mas o trabalho continua nulo se for perpendicular ao deslocamento.
- Em uma rampa, o peso pode realizar trabalho porque há componente paralela à trajetória.
