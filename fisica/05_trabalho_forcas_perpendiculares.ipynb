# Trabalho de Forças Perpendiculares ao Deslocamento
### Trabalho, energia e suas transformações

**Objetivo:** Reconhecer quando uma força não realiza trabalho porque é perpendicular ao deslocamento.

### Como usar este notebook

Execute as células na ordem, de cima para baixo. As células Markdown apresentam os conceitos; as células de código fazem as contas e geram gráficos com **NumPy** e **Matplotlib**.

> **Convenção:** neste material, usamos principalmente $W$ para trabalho, medido em joules (J). Algumas fontes usam $\tau$, mas esse símbolo também é muito utilizado para torque. O trabalho é uma grandeza escalar.

## 1. Conceito principal

O trabalho de uma força constante é $W=Fd\cos\theta$. Se a força é perpendicular ao deslocamento, $\theta=90^\circ$, e $\cos90^\circ=0$:

$$
\boxed{W=Fd\cos90^\circ=0\,\mathrm{J}}
$$

Em um bloco deslocado horizontalmente sobre uma superfície horizontal, o peso e a normal costumam ser verticais; portanto, ambos são perpendiculares ao deslocamento horizontal.

## 2. Exemplo resolvido passo a passo

Considere $d=10\,\mathrm{m}$ e uma força perpendicular de módulo $12\,\mathrm{N}$. O valor exato do módulo não altera a conclusão:

$$
W=12\cdot10\cdot\cos90^\circ=\boxed{0\,\mathrm{J}}
$$

A tabela informa o ângulo de $90^\circ$; o módulo de $12\,\mathrm{N}$ é adotado aqui apenas para ilustrar.

## 3. Conferindo o cálculo com Python

A próxima célula implementa a fórmula usando as grandezas e unidades indicadas no exemplo.

```python
import math

F = 12.0  # N; valor ilustrativo
d = 10.0  # m
theta_deg = 90.0
W = F * d * math.cos(math.radians(theta_deg))
print(f"cos(90°) = {math.cos(math.radians(theta_deg)):.1f}")
print(f"Trabalho = {W:.1f} J")
```

## 4. Visualização

O gráfico abaixo ajuda a interpretar a relação física, além de conferir os valores numéricos.

```python
import matplotlib.pyplot as plt
import numpy as np

angulos = np.linspace(0, 180, 181)
W_ang = F * d * np.cos(np.deg2rad(angulos))
plt.figure(figsize=(8, 4.5))
plt.plot(angulos, W_ang)
plt.scatter([90], [0], zorder=3, label="Força perpendicular: W = 0")
plt.axhline(0, linewidth=0.8)
plt.xlabel("Ângulo entre força e deslocamento (graus)")
plt.ylabel("Trabalho (J)")
plt.title("Em 90°, o trabalho é nulo")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## 5. Como interpretar o resultado

Uma força pode ser grande e, ainda assim, realizar trabalho nulo quando não possui componente na direção do deslocamento. Isso não significa que a força seja inexistente; significa apenas que ela não transfere energia por trabalho nesse deslocamento.

## 6. Exercício de fixação

Um corpo percorre $7\,\mathrm{m}$ horizontalmente. Qual é o trabalho do peso durante esse trecho, considerando o peso vertical?

Tente prever o resultado antes de executar a célula seguinte.

```python
F_peso_ex = 20.0
d_ex = 7.0
theta_ex = 90.0
W_peso_ex = F_peso_ex * d_ex * math.cos(math.radians(theta_ex))
print(f"Trabalho do peso = {W_peso_ex:.1f} J")
```

## 7. Resumo

- Se $\theta=90^\circ$, então $W=0$.
- Em deslocamento horizontal, peso e normal não realizam trabalho se mantiverem direção vertical.
- A conclusão depende da geometria entre força e deslocamento.
