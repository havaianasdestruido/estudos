# Conservação da Energia Mecânica: Comparação entre A e B
### Trabalho, energia e suas transformações

**Objetivo:** Comparar energia potencial e cinética em dois pontos de uma trajetória sem atrito.

### Como usar este notebook

Execute as células na ordem, de cima para baixo. As células Markdown apresentam os conceitos; as células de código fazem as contas e geram gráficos com **NumPy** e **Matplotlib**.

> **Convenção:** neste material, usamos principalmente $W$ para trabalho, medido em joules (J). Algumas fontes usam $\tau$, mas esse símbolo também é muito utilizado para torque. O trabalho é uma grandeza escalar.

## 1. Conceito principal

Em um sistema ideal sem atrito, a energia mecânica tem o mesmo valor nos pontos A e B:

$$
E_{mA}=E_{mB},\qquad E_c+E_p=\text{constante}
$$

Se o ponto B estiver mais baixo que A, a energia potencial gravitacional em B será menor. A energia cinética será maior, desde que nenhuma outra energia mude e a energia mecânica se conserve.

## 2. Exemplo resolvido passo a passo

Exemplo ilustrativo: um corpo começa em $h_A=5\,\mathrm{m}$, com velocidade $v_A=2\,\mathrm{m/s}$, e chega a $h_B=2\,\mathrm{m}$. Considere $g=10\,\mathrm{m/s^2}$.

$$
v_B^2=v_A^2+2g(h_A-h_B)=4+60=64
$$
$$
\boxed{v_B=8\,\mathrm{m/s}}
$$

A tabela original fornece a relação geral, sem valores específicos; estes números são usados para demonstrá-la.

## 3. Conferindo o cálculo com Python

A próxima célula implementa a fórmula usando as grandezas e unidades indicadas no exemplo.

```python
g = 10.0
h_A, v_A = 5.0, 2.0
h_B = 2.0
v_B = (v_A**2 + 2*g*(h_A-h_B))**0.5
Ep_A, Ec_A = g*h_A, 0.5*v_A**2  # energia por unidade de massa
Ep_B, Ec_B = g*h_B, 0.5*v_B**2
print(f"Velocidade em B: {v_B:.1f} m/s")
print(f"A: Ep/m = {Ep_A:.1f} J/kg; Ec/m = {Ec_A:.1f} J/kg; Em/m = {Ep_A+Ec_A:.1f} J/kg")
print(f"B: Ep/m = {Ep_B:.1f} J/kg; Ec/m = {Ec_B:.1f} J/kg; Em/m = {Ep_B+Ec_B:.1f} J/kg")
```

## 4. Visualização

O gráfico abaixo ajuda a interpretar a relação física, além de conferir os valores numéricos.

```python
import matplotlib.pyplot as plt

pontos = ["A", "B"]
Ep = [Ep_A, Ep_B]
Ec = [Ec_A, Ec_B]
plt.figure(figsize=(7.5, 4.5))
plt.bar(pontos, Ep, label="Energia potencial por kg")
plt.bar(pontos, Ec, bottom=Ep, label="Energia cinética por kg")
plt.ylabel("Energia por unidade de massa (J/kg)")
plt.title("A energia mecânica permanece constante")
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.show()
```

## 5. Como interpretar o resultado

A queda de altura de $3\,\mathrm{m}$ reduz $E_p/m$ em $30\,\mathrm{J/kg}$, e a energia cinética por unidade de massa aumenta exatamente pelo mesmo valor. A soma continua em $54\,\mathrm{J/kg}$.

## 6. Exercício de fixação

Um corpo sem velocidade inicial desce de $8\,\mathrm{m}$ para $3\,\mathrm{m}$, sem atrito. Com $g=10\,\mathrm{m/s^2}$, calcule a velocidade final.

Tente prever o resultado antes de executar a célula seguinte.

```python
h_A_ex, h_B_ex, g_ex = 8.0, 3.0, 10.0
v_ex = (2*g_ex*(h_A_ex-h_B_ex))**0.5
print(f"Velocidade final = {v_ex:.2f} m/s")
```

## 7. Resumo

- A energia mecânica é conservada em um sistema ideal sem atrito.
- Menor altura significa menor energia potencial gravitacional.
- A energia cinética aumenta o suficiente para manter a soma constante.
